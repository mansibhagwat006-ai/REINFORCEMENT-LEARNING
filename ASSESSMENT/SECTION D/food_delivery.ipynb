```python
# Simple Food Delivery Reinforcement Learning Episode

# Initial state
state = {
    "location": "Restaurant",
    "reward": 0,
    "steps": 0
}

# Sequence of actions taken by the agent
actions = [
    ("move_to_customer", 10),
    ("pick_up_food", 20),
    ("move_to_destination", 15),
    ("deliver_food", 50)
]

print("Food Delivery RL Episode")
print("-" * 40)

# Perform each action
for action, reward in actions:
    state["steps"] += 1
    state["reward"] += reward

    # Update location based on action
    if action == "move_to_customer":
        state["location"] = "Customer"
    elif action == "pick_up_food":
        state["location"] = "Customer"
    elif action == "move_to_destination":
        state["location"] = "Destination"
    elif action == "deliver_food":
        state["location"] = "Delivered"

    print(
        f"Step {state['steps']}: "
        f"Action = {action}, "
        f"Reward = {reward}, "
        f"Total Reward = {state['reward']}"
    )

# Final episode summary
print("\nEpisode Summary")
print("-" * 40)
print(f"Final Location : {state['location']}")
print(f"Total Reward   : {state['reward']}")
print(f"Total Steps    : {state['steps']}")
```
